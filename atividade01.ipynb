## Redimensionamento - Classe A:
- **Imagem selecionada**: cubo2.jpeg

- **Motivação**: Selecionei essa imagem por ser a de menor distância entre o objeto,além de ter a melhor iluminação, facilitando a visualização da perda de qualidade da imagem.

```python
# Lendo nova imagem e convertendo para RGB
cubo2 = cv.imread('cubo2.jpeg')
cubo2 = cv.cvtColor(cubo2, cv.COLOR_BGR2RGB)

# Obtenção da height (Altura) e width (Largura)
# para redimensionamento proporcional
h, w = cubo2.shape[:2]

# Aplicação do método de redimensionamento
cubo2_50 = cv.resize(cubo2, (w//2, h//2))
cubo2_20 = cv.resize(cubo2, (w//5, h//5))

# Criação de uma nova figure + 1 linha e 3 colunas de axes
fig, ax = plt.subplots(1,3, figsize=(12,12))

# ------------------------------
# Colocando as imagens em cada axe + títulos para melhor análise
ax[0].imshow(cubo2)
ax[0].set_title("Original")

ax[1].imshow(cubo2_50)
ax[1].set_title("Redução de 20%")

ax[2].imshow(cubo2_20)
ax[2].set_title("Redução de 50%")
# ------------------------------

# Laço para tirar os eixos x e y de cada axe
for a in ax:
  a.axis("off")

# Pra mostrar a figure criada :)
plt.show()
```

_______________________________________________________________________________
## **Observações Classe A**:
Percebe-se que as imagens foram perdendo a qualidade da esquerda para a direita. Isso se deve ao processo de redimensionamento, que diminuiu a quantidade de pixels na imagem, diminuindo também o nível de detalhamento.

## **Consequências em visão computacional**:
Para essa classe em específico, a maior consequência seria no processo de detecção de falhas presentes no cubo, já que seria mais difícil de ser enxergar falhas devida a perda de detalhes na imagem.

## Conversão de espaço de cor - Classe A:
- **Imagem selecionada**: cubo3.jpeg

- **Motivação**: Selecionei essa imagem por ter tirado ela numa área com maior diversidade de cores. Assim, fica interessante observar como o cubo mágico se comporta em relação as cores ao redor.

```python
# Lendo imagens
cubo3 = cv.imread("cubo3.jpeg")

# Convertendo para cada espaço de cor pedido
cubo3_rgb = cv.cvtColor(cubo3, cv.COLOR_BGR2RGB)
cubo3_hsv = cv.cvtColor(cubo3, cv.COLOR_BGR2HSV)
cubo3_gray = cv.cvtColor(cubo3, cv.COLOR_BGR2GRAY)

# Mais uma vez, criando figures + axes + preenchendo títulos
fig, ax = plt.subplots(1,3, figsize=(12,12))

ax[0].imshow(cubo3_rgb)
ax[0].set_title("RGB")

ax[1].imshow(cubo3_hsv)
ax[1].set_title("HSV")

# A adição da flag cmap foi a maneira de converter para gray corretamente.
ax[2].imshow(cubo3_gray, cmap='gray')
ax[2].set_title("GRAY")

for a in ax:
  a.axis('off')

plt.show()
```

_______________________________________________________________________________
## **Diferenças visuais Classe A**:

*   **RGB**: Nada fora do comum, é a escala do dia a dia, o cubo mágico é um bom destaque entre todo o resto.

*   **HSV**: Pensei que cada cor do cubo fosse ficar representada de uma forma diferente, mas estão todas bem parecidas. A imagem parece ter alguns artefatos. O mais interessante é como as cores ficara. O cinza do concreto é roxo na parte mais iluminada, mas é azul escuro na parte com sombra.

* **GRAY**: Mais uma vez, nada fora do comum, as cores da imagem passaram a ser representadas apenas por valores de 0 a 255 na escala cinza, dando a aparência de que a imagem está em preto ou branco (sei que na verdade é  preto, branco ou cinza).




## **Como eu usaria cada escala em um projeto real**:


*   **RGB**: Eu aproveitaria a visualização das cores para averiguar se a coloração do cubo estaria com o esperado na montagem, sem que houvesse quadrados de uma cor de um lado que eles não deveriam estar.

*   **HSV**: Como o HSV deixa as cores uniformes, seria bom para averiguar a uniformidade de cada quadrado, garantindo que todos estejam no mesmo tom. Provavelmente isso seria possível em RGB, mas seria mais preciso em HSV.

* **GRAY**: Já que a escala de cinza pode ser usada para operações mais simples, seria bom fazer o uso dela para detectar assimetrias no cubo, verificando se tem alguma parte torta, ou algum quadrado saindo do lugar.

## Quantização - Classe A
- **Imagem selecionada**: cubo5.jpeg

- **Motivação**: A imagem destaca bem o cubo, tem uma iluminação interessante. Quero ver o quão distorcida a imagem vai ficar quando a quantidade de bits disponíveis for diminuindo.

```python
# Aqui a leitura é feita passando o 0 como um parâmetro, o que faz
# A imagem ser lida em apenas um canal de cor, facilitando o processo.
cubo5 = cv.imread("cubo5.jpeg", 0)

# Divisão do nível de intensidade e multiplicação para aproximar o resultado
# Do nível anterior

q64 = (cubo5 // 4) * 4
q32 = (cubo5 // 8) * 8
q2 = ((cubo5 > 127) * 255).astype(np.uint8)

# Aplicando lógica anterior para mostrar imagens

fig, ax = plt.subplots(1,4, figsize=(15,15))

ax[0].imshow(cubo5, cmap='gray')
ax[0].set_title("256 Níveis")

ax[1].imshow(q64, cmap='gray')
ax[1].set_title("64 Níveis")

ax[2].imshow(q32, cmap='gray')
ax[2].set_title("32 Níveis")

ax[3].imshow(q2, cmap='gray')
ax[3].set_title("2 Níveis")

for a in ax:
  a.axis('off')

plt.show()

```

_______________________________________________________________________________
## Observações:

*   **Mudanças visuais**: Percebe-se como a imagem fica cada vez mais parecida com um desenho simples. As sombras somem e o cubo fica mais bidimensional, provavelmente por conta do ângulo.

*   **Versão mais útil para o dataset e para que fim**: Depois de testar as outras imagens, percebi que a versão mais útil poderia ser a de 2 níveis, já que elas destacam perfeitamente os quadrados coloridos. Em um ambiente de iluminação ideal e, com um cubo do modelo apresentado, a imagem de 2 níveis poderia auxiliar na análise do alinhamento dos quadrados coloridos, que ficariam brancos.

## Salvando em outros formatos - Classe A
**Imagem selecionada**: cubo1.jpeg

**Motivação**: Como a imagem é a mais borrada, gostaria de ver como ela vai se alterar quando eu salvar em outros formatos.

___

 **Formatos**:
- JPEG com compressão
- PNG sem perdas



```python
# Leitura de imagem
cubo1 = cv.imread('cubo1.jpeg')
cubo1 = cv.cvtColor(cubo1, cv.COLOR_BGR2RGB)

# Esse método recebe 2 parâmetros:
# 1. O nome do novo arquivo que vai ser criado, com extensão correspondente
# 2. A imagem a ser convertida
cv.imwrite("cubo1_1.png", cubo1)

# OBSERVAÇÃO: Meu celular salvou tudo em jpeg, então o que eu vou fazer
# é salvar novamente comprimindo a imagem.

# A lógica dos parâmetros é  a mesma de antes, com excessão do 10
# Ele representa o valor da qualidade numa escala de 0 100
# Quanto menor o valor, maior é  a compressão
cv.imwrite("cubo1_2.jpg", cubo1, [cv.IMWRITE_JPEG_QUALITY, 10])

# Método utilizado para recuperar as imagens
# a divisão converte o valor para kb
original = os.path.getsize('cubo1.jpeg') / 1024
png_size = os.path.getsize('cubo1_1.png') / 1024
jpg_size = os.path.getsize('cubo1_2.jpg') / 1024

print(f"Tamanho do original: {original:.2f} KB")
print(f"Tamanho do PNG: {png_size:.2f} KB")
print(f"Tamanho do JPEG: {jpg_size:.2f} KB")

# Demonstração visual:

fig, ax = plt.subplots(1,3, figsize=(12,12))

img_png = cv.imread('cubo1_1.png')
img_jpg = cv.imread('cubo1_2.jpg')

ax[0].imshow(cubo1)
ax[0].set_title("JPEG Original")

ax[1].imshow(img_jpg)
ax[1].set_title("JPG com perdas")

ax[2].imshow(img_png)
ax[2].set_title("PNG sem perdas")

for a in ax:
  a.axis('off')

plt.show()
```

_______________________________________________________________________________
## Observações:

*   **Tamanhos registrados**:

| Formatos | KB  | MB   
|:--------:|:--------:|:--------:|
|Original          |15.87          |0.0155          |                    
|JPEG com perdas          |13.52          |0.0132          |                    
|PNG sem perdas          |197.62          |0.154          |                  


*   **Diferenças visuais**: Da primeira para a segunda, a imagem ficou bem mais distorcida e granulada. A última imagem não parece ser diferente visualmente da primeira, mas o seu tamanho cresceu muito.


* **Que impacto isso teria em PDI**: Caso houvesse uma compressão do nível da segunda imagem, todo o pipeline de processamento seria afetado. Fazer operações de processamento de imagens pode depender muito da qualidade da imagem, dependendo do contexto. Por exemplo, caso a compressão fosse ainda maior, poderia ser impossivel de se detectar falhas em um cubo durante um processo detecção.

## Salvando em outros formatos - Classe B
**Imagem selecionada**: carta3.jpeg

**Motivação**: A imagem apresenta vários elementos presentes, o que deve facilitar o avistamento de mudanças nos diferentes formatos.
___

 **Formatos**:
- JPEG com compressão
- PNG sem perdas



```python
# Replicando lógica anterior
carta3 = cv.imread('carta3.jpeg')
carta3 = cv.cvtColor(carta3, cv.COLOR_BGR2RGB)

cv.imwrite("carta3_1.png", carta3)

cv.imwrite("carta3_2.jpg", carta3, [cv.IMWRITE_JPEG_QUALITY, 10])

original = os.path.getsize('carta3.jpeg') / 1024
png_size = os.path.getsize('carta3_1.png') / 1024
jpg_size = os.path.getsize('carta3_2.jpg') / 1024

print(f"Tamanho do original: {original:.2f} KB")
print(f"Tamanho do PNG: {png_size:.2f} KB")
print(f"Tamanho do JPEG: {jpg_size:.2f} KB")

fig, ax = plt.subplots(1,3, figsize=(16,16))

img_png = cv.imread('carta3_1.png')
img_jpg = cv.imread('carta3_2.jpg')

ax[0].imshow(carta3)
ax[0].set_title("JPEG Original")

ax[1].imshow(img_jpg)
ax[1].set_title("JPG com perdas")

ax[2].imshow(img_png)
ax[2].set_title("PNG sem perdas")

for a in ax:
  a.axis('off')

plt.show()
```
